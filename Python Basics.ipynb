```python
#HCF of Two no.
def hcf_of_num(a,b):
        if a>b:
           smallest=b
        else:
              smallest=a
        for i in range(1,smallest+1):
            if a%i==0 and b%i==0:
                hcf=i
        return hcf
num1=int(input("Enter fst nmbr:"))
num2=int(input("Enter scnd nmbr:"))
print("HCF is:",hcf_of_num(num1,num2))
         
            
            
```
